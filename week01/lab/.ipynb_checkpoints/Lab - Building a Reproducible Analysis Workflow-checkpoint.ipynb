## The reproducible analysis workflow

1. **Question:** What do we want to learn?
2. **Data:** Which observations and features can help answer it?
3. **Evidence:** What should we calculate and display to answer the question?
4. **Check:** Does the result make sense in context? Do a few original records support the meaning of the calculation?
5. **Conclusion:** What claim is supported?
6. **Limitation:** What should we avoid concluding?

**Which observations and features will you use? Explain relevant units, any filtering, and how you will handle missing or not-applicable values using the dictionary. For a percentage, define the denominator.**

Your answer: I’m going to be using all 40,034 players, comparing Gender (Male/Female) with AvgSessionDurationMinutes, which is measured in minutes. ### Step 3: Evidence

Write the code that answers your question. Break the analysis into separate cells for individual operations, as in the lecture, and add cells as needed. Display intermediate results where useful and the final result that answers the question. Explain any new calculated features. You may display a grouped summary directly without assigning it to a variable.


```python
# Prepare, filter, or calculate features as needed; display useful intermediate results.
gaming_df["Gender"].value_counts()
```

```python
# Calculate and display the evidence that answers your question. Add cells as needed.
(
    gaming_df.groupby("Gender")["AvgSessionDurationMinutes"]
    .agg(["count", "mean", "median", "min", "max"])
    .sort_values("mean", ascending=False)
)
```

**Which specific value or comparison answers your question? Explain how your calculations connect the selected data to that result.**

Your answer: Female players averaged 94.98 minutes per session, and male players averaged 94.66 minutes. The difference is only about 0.32 minutes, or roughly 19 seconds, so female players were slightly higher. There are 16,075 female players and 23,959 male players, but the two averages are almost the same.

## Investigation 2

### Step 1: Question

State a specific question that your dataset can answer. Define any important groups or comparisons.


**What is your research question 2?**

Your answer: Do players of different game genres differ in average total play time per week?


### Prediction — before calculating


**What do you predict the result will be, and why?**

Your answer: I predict RPG players will have the highest average weekly play hours. RPGs usually have quests, leveling up, and collecting items, so players might keep coming back and put in more total hours each week.


### Step 2: Data


Reuse the DataFrame and initial inspection from Investigation 1. Identify the observations and features relevant to this question; inspect additional features only if needed.


**Which observations and features will you use? Explain relevant units, any filtering, and how you will handle missing or not-applicable values using the dictionary. For a percentage, define the denominator.**

Your answer: For this question, I’m using GameGenre, SessionsPerWeek, and AvgSessionDurationMinutes. GameGenre has five categories: Action, RPG, Strategy, Sports, and Simulation. I’m calculating weekly play hours as SessionsPerWeek × AvgSessionDurationMinutes ÷ 60, so the new feature is measured in hours per week. Both numeric columns have no missing values, so I’m keeping all 40,034 players. The group counts show how many players are in each genre. This is a mean comparison, not a percentage, so there’s no percentage denominator. I’m not using PlayerID or PlayTimeHours.


### Step 3: Evidence

Write the code that answers your question. Break the analysis into separate cells for individual operations, as in the lecture, and add cells as needed. Display intermediate results where useful and the final result that answers the question. Explain any new calculated features. You may display a grouped summary directly without assigning it to a variable.


```python
# Prepare, filter, or calculate features as needed; display useful intermediate results.
gaming_df["WeeklyPlayHours"] = (
    gaming_df["SessionsPerWeek"] * gaming_df["AvgSessionDurationMinutes"] / 60
)

gaming_df[["GameGenre", "SessionsPerWeek", "AvgSessionDurationMinutes", "WeeklyPlayHours"]].head(10)
```

```python
# Calculate and display the evidence that answers your question. Add cells as needed.
(
    gaming_df.groupby("GameGenre")["WeeklyPlayHours"]
    .agg(["count", "mean", "median", "min", "max"])
    .sort_values("mean", ascending=False)
    .round(2)
)
```

**Which specific value or comparison answers your question? Explain how your calculations connect the selected data to that result.**

Your answer: Strategy players averaged 15.19 hours per week, followed by Sports at 15.03, Simulation at 14.98, Action at 14.89, and RPG at 14.72. The difference between the highest and lowest genre is only about 0.47 hours, or roughly 28 minutes. This answers the question because WeeklyPlayHours combines sessions per week with average session length, so it compares total weekly play time. The genre counts are all around 8,000 players, so the averages are based on similar group sizes.

### Step 5: Conclusion


**Answer your research question in a sentence using the numerical evidence. How does it compare with your prediction?**

Your answer: Strategy players had the highest average weekly play time at 15.19 hours, while RPG players had the lowest at 14.72 hours. The difference is only about 0.47 hours, or roughly 28 minutes, so all five genres are very close. I predicted RPG would be the highest, but it was actually the lowest, so my prediction was off. ### Step 6: Limitation


**What should a reader avoid concluding from this result, and why? Connect your limitation to this question and dataset.**

Your answer: A reader should not conclude that genre causes players to play more or less. ## Investigation 3

### Step 1: Question

State a specific question that your dataset can answer. Define any important groups or comparisons.


**What is your research question 3?**

Your answer: Does the in-game purchase rate differ across engagement levels?


### Prediction — before calculating


**What do you predict the result will be, and why?**

Your answer: I predict High engagement players will have the highest in-game purchase rate. Players who are more engaged are probably more invested in the game, so they may be more willing to spend money on it.


### Step 2: Data


Reuse the DataFrame and initial inspection from Investigation 1. Identify the observations and features relevant to this question; inspect additional features only if needed.


**Which observations and features will you use? Explain relevant units, any filtering, and how you will handle missing or not-applicable values using the dictionary. For a percentage, define the denominator.**

Your answer: For this question, I’m using EngagementLevel and InGamePurchases. EngagementLevel has three groups: Low, Medium, and High. InGamePurchases is coded as 0 for no and 1 for yes. I’m keeping all 40,034 players because neither column has missing values. For each engagement level, the denominator is the number of players in that level, and the numerator is the number of players with InGamePurchases = 1. I’ll compare the purchase rate as a percentage. The group counts show how many players are behind each percentage, and I’m not using PlayerID.


### Step 3: Evidence

Write the code that answers your question. Break the analysis into separate cells for individual operations, as in the lecture, and add cells as needed. Display intermediate results where useful and the final result that answers the question. Explain any new calculated features. You may display a grouped summary directly without assigning it to a variable.


```python
# Prepare, filter, or calculate features as needed; display useful intermediate results.
engagement_purchase = (gaming_df.groupby("EngagementLevel")["InGamePurchases"].agg(["count", "mean"]))
```

```python
# Calculate and display the evidence that answers your question. Add cells as needed.
engagement_purchase["PurchaseRatePercent"] = engagement_purchase["mean"] * 100

engagement_purchase.sort_values("PurchaseRatePercent", ascending=False).round(2)
```

**Which specific value or comparison answers your question? Explain how your calculations connect the selected data to that result.**

Your answer: High engagement players had the highest purchase rate at 20.65%, followed by Medium at 19.98% and Low at 19.73%. The gap between High and Low is only 0.92 percentage points. This answers the question because the denominator is the number of players in each engagement level, and the numerator is the number of players with InGamePurchases = 1. The counts are 10,336 High, 19,374 Medium, and 10,324 Low, so each percentage is based on more than ten thousand players.
